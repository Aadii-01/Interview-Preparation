## Complex Number Class

Create a python class ComplexNumber to represent complex numbers<br>
Theory: A complex number is a number that comprises a real part and an imaginary part. It is typically written in the form a+bi, where 'a' is the real part, and 'b' is the imaginary part, and 'i' is the imaginary unit 'root -1'
operation:
1. addition
2. subtraction
3. multiplication
4. division
5. comparison (== , !=)


````python
class ComplexNumber:
    def __init__(self,real=0.0,imag=0.0):
      self.imag = imag
      self.real = real
    def conjugate(self):
        imag = self.imag*-1
        print(f"{self.real} + {imag}i")


    def __add__(self,other):
      return ComplexNumber(self.real+other.real,self.imag+other.imag)
    def __sub__(self,other):
      return ComplexNumber(self.real-other.real,self.imag-other.imag)
    def __mul__(self,other):
      return ComplexNumber(self.real*other.real-self.imag*other.imag,self.real*other.imag+self.imag*other.real)
    def __truediv__(self,other):
      return ComplexNumber
    def __eq__(self,other):
      return self.real==other.real and self.imag==other.imag
    def __ne__(self,other):
      return self.real!=other.real or self.imag!=other.imag
    def __str__(self):
      if self.real==0:
        return f"{self.imag}i"
      elif self.imag<0:
        return f"{self.real}-{self.imag}i"
      return f"{self.real}+{self.imag}i"
````

## Inheritance in OOPS
followed in nature : parent -> child
- gives us major benifit of code reusability

User type -> login(), logout()
- Seller: login(), logout()
- Buyer: login(), logout()

> Works on the principle of DRY (Dont repeat yourself)

We inherit:
- non-private attributes
- non-private methods
- constructor [other magic methods]

> we cannot access private attributes or private methods of our class

> parent cannot inherit from child only child inherits from parent


````python
class User:
  def __init__(self, name, id):
    self.name = name
    self.id = id
  def login(self):
    print(f"{self.name} is logged in")
  def _logout(self):  #Using _value, you give the access as protected
    print(f"{self.name} is logged out")
````

Super keyword
- It is used to access methods of parents from our child class
- cannot be used outside
- can only acccess methods and not attributes

````python
class Student(User):
  def __init__(self, name, id, marks):
    super().__init__(name, id)
    self.marks = marks
  def study(self):
    print(f"{self.name} is studying")
  def login(self):
    print("OTP sent")
    super().login()
  def logout(self):
    self._logout()
````

````python
s1 = Student("aadi",10,100)
s1.login()
s1.logout()
````

````python
s2 = Student('max',2,80)
s2.study()
````

````python
s1.login()
````

### Types of Inheritance


1. Simple: Parent <-- Child
- non-pvt data members/methods can be accessed

````python
class Grandfather:
  def __init__(self,name, age):
    self.name = name
    self.age = age
  def show_details(self):
    return f"Name:{self.name} and Age:{self.age}"
  def speak(self):
    return "Grandfather speaks wisely"

class Father(Grandfather):
  def __init__(self, name, age, occupation):
    super().__init__(name, age)
    self.occupation = occupation
  def show_occupation(self):
    return f"Occupation:{self.occupation}"
  def speak(self):
    return "Father speaks carefully"
````

2. Hierarchial Inheritance:
```
Parent
 ^  ^
 |  |
 C1 C2
 ```

````python
class Grandfather:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        return f'Name: {self.name}, Age: {self.age}'

    def speak(self):
        return 'Grandfather speaks wisely.'

class Child1(Grandfather):
    def __init__(self, name, age, hobby):
        super().__init__(name, age)
        self.hobby = hobby

    def show_hobby(self):
        return f'Hobby: {self.hobby}'

    def speak(self):
        return 'Child1 speaks enthusiastically.'

class Child2(Grandfather):
    def __init__(self, name, age, favorite_subject):
        super().__init__(name, age)
        self.favorite_subject = favorite_subject

    def show_favorite_subject(self):
        return f'Favorite Subject: {self.favorite_subject}'

    def speak(self):
        return 'Child2 speaks thoughtfully.'
````

````python

# Creating objects of Child1 and Child2 classes
child1_obj = Child1('Alice', 20, 'Painting')
child2_obj = Child2('Bob', 22, 'Mathematics')
print(child1_obj.show_details())
print(child1_obj.show_hobby())
print(child1_obj.speak())

print(child2_obj.show_details())
print(child2_obj.show_favorite_subject())
print(child2_obj.speak())
````

3. Multilevel Inheritance
```
Grandparent
     ^
     |
   Parent
     ^
     |
   Child
```
- lower levels gets properties from above levels
- super() of each class can reah out to just the parent

````python
# Multilevel Inheritance Example

class Grandfather:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        return f'Name: {self.name}, Age: {self.age}'

    def speak(self):
        return 'Grandfather speaks wisely.'

class Father(Grandfather):
    def __init__(self, name, age, occupation):
        super().__init__(name, age)
        self.occupation = occupation

    def show_occupation(self):
        return f'Occupation: {self.occupation}'

    def speak(self):
        return super().speak()

class Child(Father):
    def __init__(self, name, age, hobby):
        super().__init__(name, age, 'Engineer')
        self.hobby = hobby

    def show_hobby(self):
        return f'Hobby: {self.hobby}'

    def speak(self):
        return super().speak()
````

````python

# Creating an object of the Child class
child_obj = Child('Charlie', 18, 'Cycling')
print(child_obj.show_details())
print(child_obj.show_occupation())
print(child_obj.show_hobby())
print(child_obj.speak())
````

- MRO( method resoluion operator)

````python
# Multiple Inheritance Example

class Father:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        return f'Name: {self.name}, Age: {self.age}'


    # def show_favorite_food(self):
    #     return "Here"

    def speak(self):
        return 'Father speaks carefully.'

class Mother:
    def __init__(self, name, favorite_food):
        self.name = name
        self.favorite_food = favorite_food

    def show_favorite_food(self):
        return f'Favorite Food: {self.favorite_food}'

    def speak(self):
        return 'Mother speaks lovingly.'

class Child(Mother,Father):
    def __init__(self, name, age, favorite_food, hobby):
        Father.__init__(self, name, age)
        Mother.__init__(self, name, favorite_food)
        self.hobby = hobby

    def show_hobby(self):
        return f'Hobby: {self.hobby}'

    # def speak(self):
    #     return 'Child speaks excitedly.'
````

## Abstraction
> Abstract art -> which have hidden meaning

Process of hiding complex implementations & detail, showing only essential features

from abc import ABC, abstractmethod

we have to ensure 2 things:
1. inherit abc class
2. should have an abstract method

````python
from abc import ABC,abstractmethod
````

````python
class User(ABC):
    # def __init__

    @abstractmethod
    def login(self):
        pass

    def logout(self):
        print("logout")

    @abstractmethod
    def auth(self):
        pass
````

````python
class Buyer(User):

    def login(self):
        if(self.auth):
            print("Logging in User")

    def checkObject(self,link):
        print(link)

    def auth(self):
        return True
````

````python
class Seller(User):

    def login(self):
        if(self.auth):
            print("Logging in Seller")
    def checkObject(self,link):
        print(link)

    def auth(self):
        print("OTP")
        return True
````
